# Lab 3: momentum on the Rosenbrock valley

The Rosenbrock function f(x, y) = (1 - x)^2 + 100 (y - x^2)^2 has a curved, narrow valley with its minimum at (1, 1).

```python
import numpy as np

def rosen_grad(p):
    x, y = p
    return np.array([-2 * (1 - x) - 400 * x * (y - x**2), 200 * (y - x**2)])
```

```python
p = np.array([-1.5, 2.0])
v = np.zeros(2)
lr, beta = 1e-3, 0.9
for step in range(5000):
    v = beta * v + rosen_grad(p)
    p = p - lr * v
print(p)
```

```python
import math

def cosine_lr(step, total, base_lr, min_lr=0.0):
    progress = step / total
    return min_lr + 0.5 * (base_lr - min_lr) * (1 + math.cos(math.pi * progress))
```

Without momentum the same learning rate needs far more steps to reach the minimum, because the iterate keeps bouncing across the valley floor.
