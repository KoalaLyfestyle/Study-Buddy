# Lab 5: Adam from scratch

We implement the Adam update by hand and compare it with plain gradient descent on a badly scaled quadratic.

```python
def adam_step(theta, g, m, v, t, lr=1e-3, b1=0.9, b2=0.999, eps=1e-8):
    m = b1 * m + (1 - b1) * g
    v = b2 * v + (1 - b2) * g * g
    m_hat = m / (1 - b1**t)
    v_hat = v / (1 - b2**t)
    return theta - lr * m_hat / (v_hat**0.5 + eps), m, v
```

```python
import numpy as np

scales = np.array([1.0, 100.0])
def grad(theta):
    return scales * theta
```

Adam rescales each coordinate by its own gradient history, so the steep coordinate and the flat coordinate move at similar speeds.
