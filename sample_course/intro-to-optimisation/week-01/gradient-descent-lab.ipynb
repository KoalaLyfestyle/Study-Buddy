# Lab 1: gradient descent on a parabola

We minimise f(x) = (x - 3)^2, whose minimum is at x = 3.

```python
def grad(x):
    return 2 * (x - 3)

x = 0.0
for step in range(50):
    x = x - 0.1 * grad(x)
print(round(x, 4))
```

With a learning rate of 0.1 the iterate converges to 3 within fifty steps.
